# AI Job Hunt Tracker

## Exploratory Data Analysis (EDA)

### Objective

The purpose of this notebook is to analyze the HTML structure of multiple job portals before implementing the Playwright-based job scraper.

This notebook documents:

- Website structure
- Job title selector
- Company selector
- Requirements selector
- Responsibilities selector
- Scraping challenges
- Fallback strategy

## Data Collection & Production Architecture Note

**Prototyping & EDA:** Playwright was used to automate browser interactions, analyze dynamic DOM structures, and extract job listing data during the exploratory phase.

**Production Deployment:** To optimize memory usage, lower cold-start latency, and eliminate browser dependency crashes on cloud infrastructure (Render), the production backend uses an HTTP parsing approach via axios and cheerio. This preserves the exact DOM selectors and schema established in this notebook.
